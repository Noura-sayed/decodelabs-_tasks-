## Product Performance
Insight: Sales are balanced across products.

```
product_counts = df['Product'].value_counts().sort_values()

plt.figure(figsize=(8,5))
product_counts.plot(kind='barh')
plt.title('Product demand is balanced, with Printers leading sales volume')
plt.xlabel('Orders')
plt.show()
```

## Payment Behavior
Insight: Online payments slightly lead.

```
payment_counts = df['PaymentMethod'].value_counts().sort_values()

plt.figure(figsize=(8,5))
payment_counts.plot(kind='barh')
plt.title('Customers slightly prefer online payments')
plt.xlabel('Orders')
plt.show()
```

## Order Status
Recommendation: Investigate delivery and quality issues.

```
status_percent = df['OrderStatus'].value_counts(normalize=True).sort_values()*100

plt.figure(figsize=(8,5))
status_percent.plot(kind='barh')
plt.title('More than 40% of orders are cancelled or returned')
plt.xlabel('Percentage')
plt.show()
```

## Coupon Effectiveness
Recommendation: Promote free shipping campaigns.

```
coupon_avg = df.groupby('CouponCode')['TotalPrice'].mean().sort_values()

plt.figure(figsize=(8,5))
coupon_avg.plot(kind='barh')
plt.title('FREESHIP generates the highest average order value')
plt.xlabel('Average Total Price')
plt.show()
```

## Referral Source
Recommendation: Increase Facebook ad spending.

```
referral_avg = df.groupby('ReferralSource')['TotalPrice'].mean().sort_values()

plt.figure(figsize=(8,5))
referral_avg.plot(kind='barh')
plt.title('Facebook customers spend more, Instagram drives traffic')
plt.xlabel('Average Spending')
plt.show()
```

## Weekend Analysis
Insight: Weekdays slightly outperform weekends.

```
weekend_avg = df.groupby('IsWeekend')['TotalPrice'].mean()
weekend_avg.index=['Weekday','Weekend']

plt.figure(figsize=(6,4))
weekend_avg.plot(kind='bar')
plt.title('Customers spend slightly more during weekdays')
plt.ylabel('Average Total Price')
plt.xticks(rotation=0)
plt.show()
```

## Cart Size
Recommendation: Use bundles and upselling.

```
sns.scatterplot(data=df, x='ItemsInCart', y='TotalPrice')
plt.title('Larger carts are associated with higher order value')
plt.show()
```
