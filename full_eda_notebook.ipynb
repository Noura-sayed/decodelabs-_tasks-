## Numerical Features Analysis

```
num_cols = ['Quantity','UnitPrice','ItemsInCart','TotalPrice']

df[num_cols].hist(figsize=(12,8), bins=30)
plt.tight_layout()
plt.show()
```

```
for col in num_cols:
    plt.figure(figsize=(7,2))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

## Product Analysis

```
product_counts = df['Product'].value_counts()

plt.figure(figsize=(8,5))
sns.countplot(data=df, y='Product',
              order=product_counts.index)
plt.title("Product Distribution")
plt.show()

product_counts
```

## Payment Method Analysis

```
payment_counts = df['PaymentMethod'].value_counts()

plt.figure(figsize=(8,5))
sns.countplot(data=df, x='PaymentMethod',
              order=payment_counts.index)
plt.title("Payment Methods")
plt.xticks(rotation=30)
plt.show()

payment_counts
```

## Order Status Analysis

```
status = df['OrderStatus'].value_counts(normalize=True) * 100

plt.figure(figsize=(8,5))
sns.countplot(data=df, x='OrderStatus',
              order=df['OrderStatus'].value_counts().index)
plt.title("Order Status")
plt.xticks(rotation=30)
plt.show()

status
```

## Coupon Analysis

```
coupon_counts = df['CouponCode'].value_counts()

plt.figure(figsize=(8,5))
sns.countplot(data=df, x='CouponCode',
              order=coupon_counts.index)
plt.title("Coupon Usage")
plt.show()

coupon_counts
```

```
coupon_avg = df.groupby('CouponCode')['TotalPrice'].mean().sort_values(ascending=False)
coupon_avg
```

## Referral Source

```
referral_counts = df['ReferralSource'].value_counts()

plt.figure(figsize=(8,5))
sns.countplot(data=df, x='ReferralSource',
              order=referral_counts.index)
plt.title("Referral Source")
plt.xticks(rotation=30)
plt.show()

referral_counts
```

## Weekend Analysis

```
weekend_avg = df.groupby('IsWeekend')['TotalPrice'].mean()
weekend_avg
```

## Relationship Analysis

```
sns.scatterplot(data=df, x='ItemsInCart', y='TotalPrice')
plt.title("Items In Cart vs Total Price")
plt.show()
```

```
corr = df.corr(numeric_only=True)

plt.figure(figsize=(8,5))
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
